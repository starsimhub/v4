## Exercise 1

**Suppose people with measles were infectious for only 5 days, with the
same β. What would R0 be? Change the model, run it, and check the R0 the
run prints. How does the epidemic change?**

R0 = β × the infectious period (β/γ) = 1.9 × 5 = 9.5, as `R0()`
confirms. People recover sooner, so fewer are infectious at once and the
peak is lower; but with R0 still well above 5, almost everyone is
infected.

```python
S, I, R = ss.states('S I R')
infection = S >> I @ 1.9*I
recovery8 = I >> R @ ss.days(8) # Infectious for 8 days, as in the tutorial
recovery5 = I >> R @ ss.days(5) # Infectious for 5 days instead
measles8 = ss.Disease(infection, recovery8, init_prev=0.001, name='measles')
measles5 = ss.Disease(infection, recovery5, init_prev=0.001, name='measles')

sim8 = ss.Sim(measles8, total_pop=100_000, stop=ss.days(50))
sim5 = ss.Sim(measles5, total_pop=100_000, stop=ss.days(50))
msim = ss.parallel({'Infectious for 8 days': sim8, 'Infectious for 5 days': sim5})
msim.plot('measles.I')
for sim in msim.sims:
    print(sim.label, '- R0 =', round(sim.diseases.measles.R0(), 1))
```

## Exercise 2

**Run the agent version of the model with 1,000 agents instead of 10,000
(each standing for 100 people). What changes in the plot, and why?**

Surprisingly little: in an epidemic that infects almost everyone, even
1,000 agents follow the same course, differing by little more than
chance. What changes is the detail: each agent stands for 100 people, so
results move in steps of 100, and chance plays a larger part (try
another `rand_seed=`). This matters most for small numbers, such as the
first or last few cases of an outbreak.

```python
S, I, R = ss.states('S I R')
infection = S >> I @ 1.9*I
recovery  = I >> R @ ss.days(8)
measles = ss.Disease(infection, recovery, init_prev=0.001)

many = ss.Sim(measles, total_pop=100_000, n_agents=10_000, stop=ss.days(50))
few  = ss.Sim(measles, total_pop=100_000, n_agents=1_000, stop=ss.days(50))
msim = ss.parallel({'10,000 agents': many, '1,000 agents': few})
msim.plot('measles.I')
```
